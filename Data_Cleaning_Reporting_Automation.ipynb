```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")
```

```python
df.head()
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
print(df.info())
print(df.describe())
```

```python
report = df.describe()

report.to_csv("Data_Report.csv")

print(report)
```

```python
sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(8,5))

sales.plot(kind="bar")

plt.title("Sales by Category")

plt.xlabel("Category")

plt.ylabel("Sales")

plt.show()
```

```python
profit = df.groupby("Category")["Profit"].sum()

plt.figure(figsize=(8,5))

profit.plot(kind="bar")

plt.title("Profit by Category")

plt.xlabel("Category")

plt.ylabel("Profit")

plt.show()
```
